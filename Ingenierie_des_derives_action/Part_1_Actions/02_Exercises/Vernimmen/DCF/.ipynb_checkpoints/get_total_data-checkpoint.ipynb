## Récupération des éléments financiers de l'entreprise Total Energie.
# L'objectif étant d'avoir le compte de résultat, bilan, flux de trésorerie pour appliquer le modèle DCF afin d'en tirer le prix d'une action.
> Ces données utilisées ont pour but de m'exercer premièrement à l'assimilation de ces données, de structurer le modèle et d'obtenir une valeur finale d'action, en déduire ensuite les raison pour lesquelles cette valeur est égale ou diffère de la valeur actuelle, ce qui sera la suite de mon apprentissage.

```python
company =yf.Ticker("TTE.PA")
```

```python
import pandas as pd

compte_de_resultat = company.income_stmt
bilan=company.balance_sheet
flux_de_tresorerie=company.cashflow

#comme les données enregistrées sont de type dataframe, je peux utiliser la fonction to.excel() de pandas pour écrire les données d'un dataframe dans un fichier excel

with pd.ExcelWriter("Total_Elements_Financiers.xlsx") as writer:
    compte_de_resultat.to_excel(writer, sheet_name = "Compte de Résultat")
    bilan.to_excel(writer, sheet_name = "Bilan")
    flux_de_tresorerie.to_excel(writer, sheet_name="Flux des Trésorerie")
```
